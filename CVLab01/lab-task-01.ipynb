```python
import cv2 as cv
import matplotlib.pyplot as plt
%matplotlib inline
from IPython.display import Image, display
```

```python
img = cv.imread('./assets/buter.jpg')
```

```python
# Encode from jpg to png bytes using opencv
_ , butterfly = cv.imencode('.png', img)
```

```python
# Disply inline image
display(Image(data=butterfly.tobytes()))
```

```python
height, width = img.shape[:2]
print("Height:", height)
print("Width:",width)
```

```python
resized_image = cv.resize(img, (width//2, height//2))
# malplotlib expect RGB image but OpenCV read BGR image so, we need to reverse the channels.
img = cv.imread('./assets/buter.jpg', 1)
img_channels_reverse = img[:, :,::-1]
plt.imshow(img)
plt.show()
plt.imshow(img_channels_reverse)
plt.show()
```

```python
rotated_image = cv.rotate(img_channels_reverse, cv.ROTATE_180)
plt.imshow(rotated_image)
plt.show()
```

```python
# flip the above image vertically
vertical_flip = cv.flip(img_channels_reverse, 0)
plt.imshow(vertical_flip)
plt.show()
```
