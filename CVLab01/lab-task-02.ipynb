- Take the image from Exercise 2. 
- Blur the image using a Gaussian blur with a kernel size of 5x5. 
- Convert the blurred image to grayscale. 
- Saved the image 

```python
import cv2 as cv
import matplotlib.pyplot as plt
%matplotlib inline
from IPython.display import Image, display
```

```python
# Load image and replicate lab-task-01 pipeline (produces vertical_flip)
img = cv.imread('./assets/buter.jpg')
img_channels_reverse = img[:, :, ::-1]  # BGR to RGB
vertical_flip = cv.flip(img_channels_reverse, 0)
# Kernal value is 51 because it is odd and greater than 0
gaussian_Blur_Img = cv.GaussianBlur(vertical_flip, (19, 19), 0)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.imshow(vertical_flip)
ax1.set_title('Vertical Flip')
ax1.axis('off')
ax2.imshow(gaussian_Blur_Img)
ax2.set_title('Gaussian Blur')
ax2.axis('off')
plt.tight_layout()
plt.show()
```

```python
# Convert the blurred image to grayscale
gray_image = cv.cvtColor(gaussian_Blur_Img, cv.COLOR_BGR2GRAY)
plt.imshow(gray_image)
plt.show()
# Save the image
cv.imwrite('./assets/gray_image.jpg', gray_image)
```
