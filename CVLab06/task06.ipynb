# Task 6: Shear (Slanted / 3D-Like Poster)

Affine shear via 2x3 matrix and `warpAffine`.

## Imports & Text Image

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('../CVLab02/assets/google.jpg')
h, w = img.shape[:2]
```

## Shear

```python
sh_x, sh_y = 0.35, 0.1  # horizontal slant stronger for poster look
M = np.array([[1, sh_x, -sh_x * h * 0.15], [sh_y, 1, -sh_y * w * 0.1]], dtype=np.float32)
sheared = cv2.warpAffine(img, M, (w, h), borderMode=cv2.BORDER_REPLICATE)

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB));      ax[0].set_title('Original'); ax[0].axis('off')
ax[1].imshow(cv2.cvtColor(sheared, cv2.COLOR_BGR2RGB)); ax[1].set_title('Sheared'); ax[1].axis('off')
plt.tight_layout(); plt.show()
```
