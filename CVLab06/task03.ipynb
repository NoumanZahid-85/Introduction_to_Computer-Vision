# Task 3: Horizontal and Vertical Flip

`cv2.flip` for mirroring (reflection).

## Imports & Load

```python
import cv2
import matplotlib.pyplot as plt

img = cv2.imread('../CVLab02/assets/buter.jpg')
```

## Flip

```python
flip_h = cv2.flip(img, 1)   # mirror left-right
flip_v = cv2.flip(img, 0)   # mirror top-bottom
flip_hv                  = cv2.flip(img, -1)  # both

fig, ax = plt.subplots(2, 2, figsize=(10, 10))
for a, im, t in zip(ax.flat, [img, flip_h, flip_v, flip_hv], ['Original', 'Horizontal', 'Vertical', 'Both']):
    a.imshow(cv2.cvtColor(im, cv2.COLOR_BGR2RGB)); a.set_title(t); a.axis('off')
plt.tight_layout(); plt.show()
```
