# Task 5: Warp for Face Alignment

Detect face, warp to a fixed canonical box (affine / similarity-style crop).

## Imports & Load

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('../CVLab02/assets/batman.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

## Detect face & affine warp

```python
cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
faces = cascade.detectMultiScale(gray, 1.1, 4)

OUT_W, OUT_H = 200, 220
# Canonical triangle: left-eye region, right-eye, chin (bottom center of box)

if len(faces) > 0:
    x, y, fw, fh = max(faces, key=lambda f: f[2] * f[3])
else:
    # Fallback demo ROI if no face (center box)
    h, w = gray.shape
    fw, fh = w // 3, h // 2
    x, y = (w - fw) // 2, (h - fh) // 2

src = np.float32([
    [x + 0.35 * fw, y + 0.4 * fh],
    [x + 0.65 * fw, y + 0.4 * fh],
    [x + 0.5 * fw, y + 0.9 * fh],
])
dst = np.float32([
    [0.35 * OUT_W, 0.35 * OUT_H],
    [0.65 * OUT_W, 0.35 * OUT_H],
    [0.5 * OUT_W, 0.85 * OUT_H],
])
M = cv2.getAffineTransform(src, dst)
aligned = cv2.warpAffine(img, M, (OUT_W, OUT_H))

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); ax[0].set_title('Input'); ax[0].axis('off')
ax[1].imshow(cv2.cvtColor(aligned, cv2.COLOR_BGR2RGB)); ax[1].set_title('Aligned face box'); ax[1].axis('off')
plt.tight_layout(); plt.show()
```
