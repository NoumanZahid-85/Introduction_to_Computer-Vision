# Task 10: Extract & Rectify Table / Region

Four corners of a skewed rectangle -> axis-aligned output (document scan).

## Imports & Document

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

doc = cv2.imread('../CVLab01/assets/image.jpg')
h, w = doc.shape[:2]

plt.imshow(cv2.cvtColor(doc, cv2.COLOR_BGR2RGB)); plt.title('Original'); plt.axis('off'); plt.show()
```

## Corners on skewed region -> rectangle

```python
# Quasi-skewed quadrilateral over the content (tune corners for your scan)
margin = 8
skew = 12
pts_src = np.float32([
    [margin + skew, margin],
    [w - margin, margin + skew // 2],
    [w - margin - skew, h - margin],
    [margin, h - margin - skew // 2],
])
OUT_W, OUT_H = 400, 300
pts_dst = np.float32([[0, 0], [OUT_W, 0], [OUT_W, OUT_H], [0, OUT_H]])

H = cv2.getPerspectiveTransform(pts_src, pts_dst)
rectified = cv2.warpPerspective(doc, H, (OUT_W, OUT_H))

vis = doc.copy()
cv2.polylines(vis, [pts_src.astype(int)], True, (0, 255, 0), 2)

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(cv2.cvtColor(vis, cv2.COLOR_BGR2RGB));     ax[0].set_title('Marked region'); ax[0].axis('off')
ax[1].imshow(cv2.cvtColor(rectified, cv2.COLOR_BGR2RGB)); ax[1].set_title('Rectified'); ax[1].axis('off')
plt.tight_layout(); plt.show()
```
