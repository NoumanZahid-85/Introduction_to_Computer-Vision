# Task 9: Crop ROI + Rescale (ID Card Size)

Select a region, resize to fixed output (e.g. photo slot on an ID).

## Imports & Load

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('../CVLab02/assets/batman.jpg')
h, w = img.shape[:2]
```

## Crop + fixed resize

```python
# ROI (x, y, w, h) - replace with detector output in practice
x0, y0, cw, ch = int(0.25 * w), int(0.15 * h), int(0.5 * w), int(0.65 * h)
roi = img[y0:y0 + ch, x0:x0 + cw]

PHOTO_W, PHOTO_H = 280, 350  # target slot
id_photo = cv2.resize(roi, (PHOTO_W, PHOTO_H), interpolation=cv2.INTER_AREA)

fig, ax = plt.subplots(1, 3, figsize=(12, 3))
ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB));       ax[0].set_title('Full'); ax[0].axis('off')
ax[1].imshow(cv2.cvtColor(roi, cv2.COLOR_BGR2RGB));       ax[1].set_title('Cropped ROI'); ax[1].axis('off')
ax[2].imshow(cv2.cvtColor(id_photo, cv2.COLOR_BGR2RGB)); ax[2].set_title(f'Rescaled {PHOTO_W}x{PHOTO_H}'); ax[2].axis('off')
plt.tight_layout(); plt.show()
```
