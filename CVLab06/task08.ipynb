# Task 8: Multiple Transforms in Sequence

Chain 3x3 matrices: overall `M = Mn @ ... @ M2 @ M1` applied once.

## Imports & Image

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('../CVLab01/assets/buter.jpg')
h, w = img.shape[:2]
```

## Build pipeline: scale, rotate, translate

```python
def scale_matrix(sx, sy):
    return np.array([[sx, 0, 0], [0, sy, 0], [0, 0, 1]], dtype=np.float32)


def rotate_matrix(deg, cx, cy):
    t = np.radians(deg)
    c, s = np.cos(t), np.sin(t)
    return np.array([[c, -s, cx - c * cx + s * cy], [s, c, cy - s * cx - c * cy], [0, 0, 1]], dtype=np.float32)


def translate_matrix(tx, ty):
    return np.array([[1, 0, tx], [0, 1, ty], [0, 0, 1]], dtype=np.float32)


cx, cy = w / 2, h / 2
S = scale_matrix(0.9, 0.9)
R = rotate_matrix(12, cx, cy)
T = translate_matrix(20, -15)
# Apply translate last on pixel: p' = T @ R @ S @ p  (scale first at origin logic)
M = T @ R @ S
out = cv2.warpPerspective(img, M, (w, h))

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); ax[0].set_title('Original'); ax[0].axis('off')
ax[1].imshow(cv2.cvtColor(out, cv2.COLOR_BGR2RGB)); ax[1].set_title('S then R then T'); ax[1].axis('off')
plt.tight_layout(); plt.show()
```
