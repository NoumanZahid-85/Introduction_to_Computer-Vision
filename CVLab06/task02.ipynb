# Task 2: Resize Inside 500x500 (Aspect Ratio)

Scale uniformly so the image fits in 500x500 without stretching; pad with black.

## Imports & Load

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('../CVLab02/assets/batman.jpg')
h, w = img.shape[:2]
FRAME = 500
```

## Fit inside frame

```python
scale = min(FRAME / w, FRAME / h)  # same scale for both axes
nw, nh = int(w * scale), int(h * scale)
resized = cv2.resize(img, (nw, nh), interpolation=cv2.INTER_AREA)

canvas = np.zeros((FRAME, FRAME, 3), dtype=np.uint8)
y0, x0 = (FRAME - nh) // 2, (FRAME - nw) // 2
canvas[y0:y0 + nh, x0:x0 + nw] = resized

fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB)); ax[0].set_title(f'Original {w}x{h}'); ax[0].axis('off')
ax[1].imshow(cv2.cvtColor(canvas, cv2.COLOR_BGR2RGB)); ax[1].set_title(f'Inside {FRAME}x{FRAME}'); ax[1].axis('off')
plt.tight_layout(); plt.show()
print('Aspect ratio kept; letterboxing fills unused area.')
```
