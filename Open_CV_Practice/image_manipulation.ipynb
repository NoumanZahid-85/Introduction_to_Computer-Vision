### Display image as grayscale(0)

### Doaplay image as color(1)

```python
# read image as color
img = cv.imread('./assets/buter.jpg', 1)
# display image
plt.imshow(img)
```

```python
# malplotlib expect RGB image but OpenCV read BGR image so, we need to reverse the channels.
img = cv.imread('./assets/buter.jpg', 1)
img_channels_reverse = img[:, :,::-1]
plt.imshow(img)
plt.show()
plt.imshow(img_channels_reverse)
plt.show()
```

### Image Atributes

```python
print("Image Shape: ",img.shape)
print("Image Size: ",img.size)
print("Image dtype: ",img.dtype)
print("Image itemsize: ",img.itemsize)
print("Image nbytes: ",img.nbytes)
print("Image ndim: ",img.ndim)
print("Image dtype: ",img.dtype)
```

### Split into BGR Components

```python
# read image as color
img_BGR = cv.imread('./assets/buter.jpg', cv.IMREAD_COLOR)
# split into BGR components
b, g, r = cv.split(img_BGR)
# display each channel
plt.figure(figsize=(20, 5))
plt.subplot(141); plt.imshow(b, cmap='gray'); plt.title('Blue Channel')
plt.subplot(142); plt.imshow(g, cmap='gray'); plt.title('Green Channel')
plt.subplot(143); plt.imshow(r, cmap='gray'); plt.title('Red Channel')         
### Merge BGR Components
img_BGR_merged = cv.merge((b, g, r))
plt.subplot(144); plt.imshow(img_BGR_merged[:, :, ::-1]); plt.title('Merged Image')
```

### Changes into HSV Color space

```python
img_HSV = cv.cvtColor(img_BGR, cv.COLOR_BGR2HSV)
h, s, v = cv.split(img_HSV)
plt.figure(figsize=(20, 5))
plt.subplot(141); plt.imshow(h, cmap='gray'); plt.title('Hue Channel')
plt.subplot(142); plt.imshow(s, cmap='gray'); plt.title('Saturation Channel')
plt.subplot(143); plt.imshow(v, cmap='gray'); plt.title('Value Channel')
plt.subplot(144); plt.imshow(img_HSV[:, :, ::-1]); plt.title('Merged Image')
plt.show()
```

```python
# Change the value of Hue
h_new = h + 20
img_HSV_new = cv.merge((h_new, s, v))
img_BGR_new = cv.cvtColor(img_HSV_new, cv.COLOR_HSV2BGR)
# Show channels
plt.figure(figsize=(20, 5))
plt.subplot(141); plt.imshow(h, cmap='gray'); plt.title('Hue Channel')
plt.subplot(142); plt.imshow(s, cmap='gray'); plt.title('Saturation Channel')
plt.subplot(143); plt.imshow(v, cmap='gray'); plt.title('Value Channel')
plt.subplot(144); plt.imshow(img_BGR_new[:, :, ::-1]); plt.title('Merged Image')
plt.show()
```

### Save the modified image

```python
final_img = cv.imwrite('./assets/buter_modified.jpg', img_BGR_new)
Image(filename='./assets/buter_modified.jpg', width=450, height=450)
```
