## Start with Open CV

### Import Libraries

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
from IPython.display import Image
```

### Display Image 

```python
Image(filename='./buter.jpg', width=450, height=450)
```

### Display image as grayscale(0)

```python
# read image as grayscale
img = cv.imread('./buter.jpg', 0)
print(img)
# display image
plt.imshow(img, cmap='gray')
plt.show()
```

### Doaplay image as color(1)

```python
# read image as color
img = cv.imread('./buter.jpg', 1)
# display image
plt.imshow(img)
```

```python
# malplotlib expect RGB image but OpenCV read BGR image so, we need to reverse the channels.
img = cv.imread('./buter.jpg', 1)
img_channels_reverse = img[:, :,::-1]
plt.imshow(img)
plt.show()
plt.imshow(img_channels_reverse)
plt.show()
```

### Image Atributes

```python
print("Image Shape: ",img.shape)
print("Image Size: ",img.size)
print("Image dtype: ",img.dtype)
print("Image itemsize: ",img.itemsize)
print("Image nbytes: ",img.nbytes)
print("Image ndim: ",img.ndim)
print("Image dtype: ",img.dtype)
```
