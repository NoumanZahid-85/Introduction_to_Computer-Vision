# Lab Task 1: Erosion and Dilation

## Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Load Image

```python
img = cv2.imread('../CVLab02/assets/batman.jpg', cv2.IMREAD_GRAYSCALE)

plt.imshow(img, cmap='gray'); plt.title('Original'); plt.axis('off'); plt.show()
```

## Erosion — Varying Kernel Sizes

```python
sizes = [3, 7, 11]
eroded = [cv2.erode(img, np.ones((k, k), np.uint8), iterations=1) for k in sizes]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle('Erosion — Kernel Size Comparison', fontweight='bold')
for ax, res, k in zip(axes, eroded, sizes):
    ax.imshow(res, cmap='gray'); ax.set_title(f'Erode k={k}'); ax.axis('off')
plt.tight_layout(); plt.show()
```

## Erosion — Varying Iterations

```python
kernel = np.ones((5, 5), np.uint8)
iters  = [1, 2, 4]
eroded_iters = [cv2.erode(img, kernel, iterations=i) for i in iters]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle('Erosion — Iteration Comparison (k=5)', fontweight='bold')
for ax, res, i in zip(axes, eroded_iters, iters):
    ax.imshow(res, cmap='gray'); ax.set_title(f'iter={i}'); ax.axis('off')
plt.tight_layout(); plt.show()
```

## Dilation — Varying Kernel Sizes

```python
dilated = [cv2.dilate(img, np.ones((k, k), np.uint8), iterations=1) for k in sizes]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle('Dilation — Kernel Size Comparison', fontweight='bold')
for ax, res, k in zip(axes, dilated, sizes):
    ax.imshow(res, cmap='gray'); ax.set_title(f'Dilate k={k}'); ax.axis('off')
plt.tight_layout(); plt.show()
```

## Dilation — Varying Iterations

```python
dilated_iters = [cv2.dilate(img, kernel, iterations=i) for i in iters]

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle('Dilation — Iteration Comparison (k=5)', fontweight='bold')
for ax, res, i in zip(axes, dilated_iters, iters):
    ax.imshow(res, cmap='gray'); ax.set_title(f'iter={i}'); ax.axis('off')
plt.tight_layout(); plt.show()
```

## Side-by-Side Comparison

```python
k = np.ones((5, 5), np.uint8)
imgs   = [img, cv2.erode(img, k, iterations=2), cv2.dilate(img, k, iterations=2)]
labels = ['Original', 'Eroded (k=5, iter=2)', 'Dilated (k=5, iter=2)']

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
fig.suptitle('Erosion vs Dilation', fontweight='bold')
for ax, im, lbl in zip(axes, imgs, labels):
    ax.imshow(im, cmap='gray'); ax.set_title(lbl); ax.axis('off')
plt.tight_layout(); plt.show()

print('Erosion  : shrinks bright regions, enlarges dark regions, removes small white noise')
print('Dilation : expands bright regions, shrinks dark regions, fills small holes')
print('Larger kernel / more iterations = stronger effect')
```
