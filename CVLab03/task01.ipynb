## Video Processing with OpenCV 

### Lab Task 01

```python
import cv2
```

```python
video_path = 'first_video.mp4'
```

```python
cap = cv2.VideoCapture(video_path)
print(type(cap))
```

```python


while True:
    # 
    ret, frame = cap.read()

    if not ret:
        break  

    # Resize Frame
    resized = cv2.resize(frame, (225, 225))

    # 2️⃣ Apply Gaussian Blur
    blurred = cv2.GaussianBlur(resized, (15, 15), 0)

    # 3️⃣ Convert to Grayscale + Apply Colormap
    gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
    colored = cv2.applyColorMap(gray, cv2.COLORMAP_JET)

    # Combine original and processed frames side-by-side
    combined = cv2.hconcat([resized, blurred, colored])

    cv2.imshow('Original | Blurred | Colored', combined)

    # Press 'q' to quit
    if cv2.waitKey(25) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
