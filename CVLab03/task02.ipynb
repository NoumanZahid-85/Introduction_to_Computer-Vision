## Video Processing with OpenCV

### Lab Task 02: Writing to Video with OpenCV

```python
import cv2
```

```python
video_path = 'first_video.mp4'
output_path = 'output_task02.mp4'
cap = cv2.VideoCapture(video_path)
```

```python
# Video writer params: combined frame = 3*225 width, 225 height
frame_width, frame_height = 675, 225
fps = int(cap.get(cv2.CAP_PROP_FPS)) or 25
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))
frame_count = 0
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break

    resized = cv2.resize(frame, (225, 225))
    blurred = cv2.GaussianBlur(resized, (15, 15), 0)
    gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
    colored = cv2.applyColorMap(gray, cv2.COLORMAP_JET)
    combined = cv2.hconcat([resized, blurred, colored])

    out.write(combined)
    frame_count += 1
    if frame_count % 50 == 0:
        print(f"Progress: {frame_count} frames written...")

cap.release()
out.release()
cv2.destroyAllWindows()
print(f"Video creation complete. Saved to '{output_path}' ({frame_count} frames, {fps} fps).")
```
