## Video Processing with OpenCV

### Lab Task 03: Writing Text on Video

```python
import cv2
```

```python
video_path = 'first_video.mp4'
output_path = 'output_task03.mp4'
cap = cv2.VideoCapture(video_path)
```

```python
frame_width, frame_height = 675, 225
fps = int(cap.get(cv2.CAP_PROP_FPS)) or 25
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (frame_width, frame_height))
frame_count = 0
font = cv2.FONT_HERSHEY_SIMPLEX
font_scale, color, thickness = 1.2, (0, 255, 255), 2  # Yellow text
```

```python
while True:
    ret, frame = cap.read()
    if not ret:
        break

    resized = cv2.resize(frame, (225, 225))
    blurred = cv2.GaussianBlur(resized, (15, 15), 0)
    gray = cv2.cvtColor(resized, cv2.COLOR_BGR2GRAY)
    colored = cv2.applyColorMap(gray, cv2.COLORMAP_JET)
    combined = cv2.hconcat([resized, blurred, colored])

    # Write frame number on each frame
    text = f"Frame: {frame_count}"
    cv2.putText(combined, text, (20, 35), font, font_scale, color, thickness)

    out.write(combined)
    frame_count += 1
    if frame_count % 50 == 0:
        print(f"Progress: {frame_count} frames written...")

cap.release()
out.release()
cv2.destroyAllWindows()
print(f"Video creation complete. Saved to '{output_path}' ({frame_count} frames). Text overlay applied.")
```
