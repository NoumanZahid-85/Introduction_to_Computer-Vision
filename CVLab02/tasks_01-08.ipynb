### `Task 1: Drawing Ellipses`

```python
import cv2 as cv
import matplotlib.pyplot as plt
%matplotlib inline
from IPython.display import Image, display
```

```python
# Load the image
image = cv.imread('./assets/buter.jpg', 1)
# Draw on BGR image (OpenCV requires contiguous array; img[:,:,::-1] creates non-contiguous view)
# Parameters: (image, center, axes, angle, startAngle, endAngle, color_BGR, thickness)
cv.ellipse(image, (800, 350), (300, 150), 0, 0, 360, (0, 0, 255), 9)   
cv.ellipse(image, (800, 350), (200, 330), 30, 0, 360, (0, 255, 0), 9)   
cv.ellipse(image, (800, 350), (250, 180), 50, 0, 360, (255, 0, 0), 9)  
 
# Convert to RGB for matplotlib display
img = image[:, :, ::-1]
plt.imshow(img)
plt.title('Drawing Ellipses')
plt.axis('off')
plt.show()

# Display in separate window
# cv.imshow("Task 1 - Ellipses", image)
# cv.waitKey(0)
# cv.destroyAllWindows()
```

### `Task 2: Converting Ellipses to Polygons`

```python
# Simple Ellipse
image_1 = cv.imread('./assets/buter.jpg', 1)
cv.ellipse(image_1, (800, 350), (300, 150), 0, 0, 360, (255, 0, 0), 9) 
# Ellipse to polygon - returns 2D array of points, NOT an image
poly_pts = cv.ellipse2Poly((800, 350), (300, 150), 0, 0, 360, 100)
image_2 = cv.imread('./assets/buter.jpg', 1).copy()
cv.polylines(image_2, [poly_pts], True, (255, 0, 0), 3)  # draw polygon on image
# Display side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
ax1.imshow(image_1[:, :, ::-1])
ax1.set_title('Ellipse')
ax1.axis('off')
ax2.imshow(image_2[:, :, ::-1])
ax2.set_title('Ellipse as Polygon')
ax2.axis('off')
plt.tight_layout()
plt.show()

    
```

### `Task 3: Filling Convex Polygons `

```python
# I Use polygonized ellipse from Task 2
poly_pts = cv.ellipse2Poly((800, 350), (300, 150), 0, 0, 360, 100)
image = cv.imread('./assets/buter.jpg', 1).copy()
# Fill the convex polygon with color (BGR: purple)
cv.fillConvexPoly(image, poly_pts, (255, 0, 255))
# Display
img_rgb = image[:, :, ::-1]
plt.imshow(img_rgb)
plt.title('Filled Convex Polygon (ellipse)')
plt.axis('off')
plt.show()
```

### `Task 4: Drawing Rectangles `

```python
image = cv.imread('./assets/buter.jpg', 1).copy()
# cv.rectangle(img, pt1, pt2, color, thickness)
cv.rectangle(image, (200, 100), (500, 350), (0, 255, 0), 8)   # green, thickness 3
cv.rectangle(image, (600, 200), (1200, 500), (255, 0, 0), 5)  # blue, thickness 5
cv.rectangle(image, (100, 400), (400, 700), (0, 0, 255), 7)   # red, thickness 2
img_rgb = image[:, :, ::-1]
plt.imshow(img_rgb)
plt.title('Drawing Rectangles')
plt.axis('off')
plt.show()
```

### `Task 5: Drawing Lines and Arrowed Lines `

```python
image = cv.imread('./assets/buter.jpg', 1).copy()
pt1, pt2 = (100, 400), (400, 300)
cv.line(image, pt1, pt2, (0, 0, 255), 8)
cv.arrowedLine(image, (500, 150), (1100, 450), (255, 0, 0), 8)
img_rgb = image[:, :, ::-1]
plt.imshow(img_rgb)
plt.title('Lines and Arrowed Lines')
plt.axis('off')
plt.show()
```

### `Task 6: Drawing Polylines `

```python
import numpy as np
image = cv.imread('./assets/buter.jpg', 1).copy()
# Define points for a polyline (e.g., Z shape)
pts = np.array([[300, 200], [500, 400], [700, 200], [900, 400], [1100, 200]], np.int32)
pts = pts.reshape((-1, 1, 2))
cv.polylines(image, [pts], False, (0, 0, 255), 8)  # isClosed=False
img_rgb = image[:, :, ::-1]
plt.imshow(img_rgb)
plt.title('Polylines')
plt.axis('off')
plt.show()
```

### `Task 7: Putting Text on Images `

```python
image = cv.imread('./assets/buter.jpg', 1).copy()
# putText(img, text, org, fontFace, fontScale, color, thickness)
cv.putText(image, 'Nouman Zahid', (50, 100), cv.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 4)
cv.putText(image, 'Large text', (50, 200), cv.FONT_HERSHEY_SIMPLEX, 2, (0, 255, 0), 3)
cv.putText(image, 'Blue bold', (500, 400), cv.FONT_HERSHEY_DUPLEX, 1.5, (255, 0, 0), 4)
cv.putText(image, 'Butterfly', (800, 150), cv.FONT_HERSHEY_SCRIPT_SIMPLEX, 2, (0, 255, 255), 2)
img_rgb = image[:, :, ::-1]
plt.imshow(img_rgb)
plt.title('Text on Image')
plt.axis('off')
plt.show()
```

### `Task 8: Additional Challenges `

```python
# Additional function 1: getTextSize - get text dimensions to center text on image
# Additional function 2: drawMarker - draw markers (cross, circle, etc.) at points
image = cv.imread('./assets/buter.jpg', 1).copy()
h, w = image.shape[:2]

# getTextSize(text, fontFace, fontScale, thickness) -> (size, baseline)
text = 'Butterfly'
(tw, th), baseline = cv.getTextSize(text, cv.FONT_HERSHEY_SIMPLEX, 2, 3)
x = (w - tw) // 2
y = (h + th) // 2
cv.putText(image, text, (x, y), cv.FONT_HERSHEY_SIMPLEX, 2, (0, 0, 0), 9)

# drawMarker(img, position, color, markerType, markerSize, thickness)
cv.drawMarker(image, (200, 300), (0, 255, 255), cv.MARKER_CROSS, 30, 3)
cv.drawMarker(image, (600, 600), (255, 0, 255), cv.MARKER_STAR, 40, 2)

img_rgb = image[:, :, ::-1]
plt.imshow(img_rgb)
plt.title('getTextSize (center text) + drawMarker (cross, star)')
plt.axis('off')
plt.show()
```
